```python
import random
import matplotlib.pyplot as plt  


simulation_params = {
    'alfa': 3,
    'beta': 2
}

pressures = []
for i in range(1000):
    # Simulazione di un valore di pressione sanguigna
    pressure_value = 100 - random.gammavariate(simulation_params['alfa'], simulation_params['beta'])
    pressures.append(pressure_value)

# Creazione del grafico
plt.figure(figsize=(10, 5))
plt.plot(pressures, label='Pressione Sanguigna Simulata', color='blue')
plt.title('Simulazione della Pressione Sanguigna')

# Ora simuliamo la frequenza cardiaca
curr_hr = 0
heart_rates = []
for i in range(1000):
    # Simulazione di un valore di frequenza cardiaca
    curr_hr = 0.9 * curr_hr + 0.1 * random.gauss(75, 3)
    heart_rates.append(curr_hr)

plt.figure(figsize=(10, 5))
plt.plot(heart_rates, label='Frequenza Cardiaca Simulata', color='red')
plt.xlabel('Campioni')
plt.ylabel('Valore')
```
